The Heart Disease Prediction Dataset is a synthetic medical dataset created for machine learning, data analysis, and predictive modeling tasks. The dataset contains 1000 patient records with multiple health-related features including glucose level, cholesterol level, blood pressure, BMI, heart rate, lifestyle habits, and family medical history.

The primary objective of this dataset is to predict the likelihood of heart disease risk using various clinical and lifestyle indicators. The target column (disease) is designed for binary classification tasks and contains two classes: Yes and No.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,confusion_matrix
```

```python
#convering the categorical data into numerical data

#Column with binary categorical data can be converted into numerical data using map function
df["gender"]=df["gender"].map({"Male":1,"Female":0})
df["smoking"]=df["smoking"].map({"Yes":1,"No":0})
df["alcohol_consumption"]=df["alcohol_consumption"].map({"Yes":1,"No":0})
df["family_history"]=df["family_history"].map({"Yes":1,"No":0})

#Columns with more than two categories can be converted into numerical data using get_dummies function
df=pd.get_dummies(df,columns=["physical_activity"],drop_first=True)
```

```python
#dividing the data into features and target
X=df.drop("disease",axis=1)
y=df["disease"]   
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
logistic=LogisticRegression(max_iter=10000)
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)

print(classification_report(y_test,y_pred))
```

```python

print(confusion_matrix(y_test,y_pred))
```
