```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from imblearn.over_sampling import SMOTE
```

```python
df=pd.read_csv('/content/drive/MyDrive/Feature Engineering and ETA/VehicleInsuranceData/aug_train.csv',encoding='latin-1')
```

```python
df.head(300)
```

```python
df=df[["Age","Region_Code","Response"]]
df=df.head(300)
# for testing we will be using 3 columns namely Region_Code Age and Response, and 300 rows only
# here Region_Code and Age with be two feature(independent varaibles) while Response will be target /dependent varaible
```

```python
df.head()
df.count()
```

```python
plt.scatter(df['Age'],df['Region_Code'],c=df['Response'])
```

```python
oversample=SMOTE()
```

```python
X,y=df_oversampled=oversample.fit_resample(df[["Age","Region_Code"]],df['Response'])
```

```python
X.shape
```

```python
y.shape
```

```python
df1=pd.DataFrame(X,columns=["Age","Region_Code"])
df2=pd.DataFrame(y,columns=['Response'])
oversample_df=pd.concat([df1,df2],axis=1)
```
