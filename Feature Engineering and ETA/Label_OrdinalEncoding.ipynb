# Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df_n=pd.DataFrame({'size':['small','medium','large','medium','small','large']})
```

```python
encoder_1=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder_1.fit_transform(df_n)
```

```python
encoder_1.transform([['small']])
```
