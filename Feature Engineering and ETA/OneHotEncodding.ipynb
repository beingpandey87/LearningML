# One Hot Encoder / Norminal Encoder

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.read_csv('/content/drive/MyDrive/Feature Engineering and ETA/HousePrice_AdvRegressionTechnique/train.csv')
df.head()
```

```python
df.columns
```

```python
df["LotShape"].value_counts()
```

Here we will be using LotShape column data to check how One hot encodding is used.

```python
encoder=OneHotEncoder()
encoder_val=encoder.fit_transform(df[["LotShape"]]).toarray()
```

```python
encoded_df=pd.DataFrame(encoder_val,columns=encoder.get_feature_names_out())
```

```python
encoded_df.head()
```

```python
pd.concat([df['LotShape'],encoded_df],axis=1)
```
