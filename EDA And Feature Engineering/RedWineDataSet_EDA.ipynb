# **Title: Wine Quality Dataset**

## Description:

This dataset contains physicochemical and sensory attributes of 1599 samples of red wine, along with a quality rating. The data was collected to analyze and predict wine quality based on its chemical properties. No missing values were present in the original dataset.

## Features:

The dataset includes the following attributes, all measured quantitatively:

- fixed acidity: The amount of tartaric acid (g/L).
- volatile acidity: The amount of acetic acid (g/L), which can lead to unpleasant vinegar-like tastes.
- citric acid: The amount of citric acid (g/L), adding freshness and flavor.
- residual sugar: The amount of sugar remaining after fermentation (g/L).
- chlorides: The salt content (g/L).
- free sulfur dioxide: The amount of SO₂ not bound to other molecules (mg/L).
- total sulfur dioxide: The total amount of SO₂ (mg/L), which acts as a preservative.
- density: The mass per unit volume (g/cm³), influenced by alcohol and sugar content.
- pH: A measure of acidity/basicity on a logarithmic scale.
sulphates: The amount of potassium sulphate (g/L), which can affect SO₂ levels.
- alcohol: The alcohol content by volume (%).
- quality: A sensory rating (0–10, though most wines are rated 3–9).
Potential Uses:

### This dataset can be used for a variety of machine learning and statistical tasks, including:

## Regression Analysis: Predicting wine quality scores based on physicochemical features.
- Classification: Categorizing wines into quality tiers (e.g., low/medium/high).
- Feature Importance: Identifying which chemical properties most influence perceived quality.
- Process Optimization: Helping winemakers adjust variables to improve quality.
- Exploratory Data Analysis: Understanding correlations between chemical properties and taste.

## Dataset Characteristics:

- Numerical features (continuous variables).
- Target variable (quality) is ordinal (categorical with inherent order).
- Suitable for regression, classification, and clustering tasks.
